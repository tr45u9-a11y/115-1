```python
listA=[]
while True:
  item=input("請輸入A倉文具，輸入'/'結束:")
  if item=="/":
    break
  listA.append(item)
listB=[]
while True:
  item=input("請輸入B倉文具，輸入'/'結束:")
  if item=="/":
    break
  listA.append(item)
listC=listA+listB
listC.append("膠水")
listC="/".join(listC)
print(f"二倉道合併清單:{listC}")
```

```python
order=[]
print("請輸入您要點的餐點,輸入'/'結束:")
while True:
  item=input()
  if item=="/":
    break
  order.append(item)
  if item=="沙拉":
    order.remove("沙拉")
  print("今日無沙拉，已自動移除")
  order.append("冰淇淋")
  order=",".join(order)
  print(f"完整訂單清單:{order}")
```

```python
import random
day=1
max_day=3
win=False
while day<=max_day and not win:
  lucky=random.randint(1,50)
  guess=int(input(f"今天第{day}天，請輸入數字:"))
  print(f"今天的幸運數字{lucky}")
  if guess == lucky:
    print("你中獎了")
    win=True
  else:
    print("你沒中獎，明天再來")
    day+=1
if not win:
  print("很遺憾3天都沒有中獎，下次再試試看")
```
